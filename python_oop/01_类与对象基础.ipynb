## 1.1 为什么需要面向对象？

在编程中，我们通常有两种组织代码的思路：

| 对比维度 | 面向过程（POP） | 面向对象（OOP） |
|---------|----------------|----------------|
| 核心思想 | 以**步骤/函数**为中心 | 以**数据与行为的封装体（对象）**为中心 |
| 代码组织 | 函数 + 全局数据 | 类把数据和操作数据的方法绑定在一起 |
| 复用方式 | 复制函数、传参 | 继承、组合、多态 |
| 适合场景 | 脚本、一次性任务 | 中大型系统、长期维护的项目 |
| 典型问题 | 数据与操作分离，改一处容易漏另一处 | 学习曲线略陡 |

举个直观的例子——描述一只狗：

- **面向过程**：用字典存数据 `dog = {"name": "旺财", "age": 3}`，再写一堆函数 `bark(dog)`、`feed(dog)` 操作它。
- **面向对象**：定义一个 `Dog` 类，狗自己会 `bark()`、会 `eat()`，数据和能力是一体的。

```mermaid

        D1[数据: name, age] --> F1[bark 函数]
        D1 --> F2[feed 函数]
        F1 -->|手动保证数据格式正确| R1[结果]
        F2 --> R1
    end
    subgraph OOP[面向对象]
        O[Dog 对象<br/>数据 + 行为一体] --> M1[dog.bark]

    end
```

## 1.3 类与对象：图纸与房子

- **类（Class）**：一张**图纸/模板**，描述这类事物有什么属性、能做什么事。
- **对象/实例（Instance）**：按图纸造出来的**具体房子**，每个实例有自己独立的数据。

```mermaid
classDiagram
    class Dog {
        +str species 类属性
        +str name
        +int age
        +bark() str
    }
    Dog <|.. 旺财 : 实例化
    Dog <|.. 大黄 : 实例化
```

一个类可以实例化出**任意多个**对象，对象之间互不干扰（它们共享方法，但各自持有自己的数据）。

## 1.4 定义第一个类

语法要点：
- `class 类名:` —— 类名约定使用**大驼峰**（`PascalCase`）
- `def __init__(self, ...)` —— **构造方法**，创建实例时自动调用，用于初始化数据
- `self` —— 指向**当前实例自身**的引用，是方法的第一个参数（约定俗成叫 `self`）

````python
class Dog:
    """一个简单的狗类"""

    def __init__(self, name, age):
        # self.xxx 叫做实例属性，绑定在具体对象上
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name}: 汪汪汪！"


# 实例化：像调用函数一样调用类名
dog1 = Dog("旺财", 3)
dog2 = Dog("大黄", 5)

print(dog1.bark())
print(dog2.bark())
print(dog1.name, dog1.age)   # 直接访问实例属性
print(type(dog1))            # <class '__main__.Dog'>
````

## 1.5 `self` 到底是什么？

`self` **不是关键字**，它只是一个约定俗成的参数名。当你写 `dog1.bark()` 时，Python 实际做的是：

```python
Dog.bark(dog1)   # 解释器把实例 dog1 作为第一个参数传进去
```

```mermaid
sequenceDiagram
    participant U as 你的代码
    participant C as 类 Dog
    participant I as 实例 dog1
    U->>I: dog1.bark()
    I->>C: 查找 bark 方法
    C-->>I: 找到函数 bark(self)
    I->>C: 自动传入 dog1 作为 self
    C-->>U: 返回 "旺财: 汪汪汪！"
```

> 💡 **对比 Java/C++**：它们用 `this`，且是**隐式**的；Python 的 `self` 是**显式**声明的第一个参数——这体现了 Python 哲学 "Explicit is better than implicit"。

````python
class Demo:
    def who_am_i(self):
        print(f"self 的内存地址: {id(self)}")


d = Demo()
print(f"d    的内存地址: {id(d)}")
d.who_am_i()          # 两个地址相同 → self 就是实例本身

# 手动演示解释器的调用方式（实际开发中不要这么写）
Demo.who_am_i(d)
````

## 1.6 类属性 vs 实例属性

| 对比维度 | 类属性 | 实例属性 |
|---------|--------|---------|
| 定义位置 | 类体内、方法之外 | 通常在 `__init__` 中，通过 `self.xxx =` |
| 归属 | 属于**类本身**，所有实例共享 | 属于**单个实例**，各自独立 |
| 内存 | 一份 | 每个实例一份 |
| 访问方式 | `类名.属性` 或 `实例.属性` | `实例.属性` |
| 典型用途 | 常量、计数器、默认值 | 每个对象特有的状态 |

**查找顺序**：访问 `obj.attr` 时，Python 先查**实例**字典，找不到再查**类**字典（再查父类，见第 3 章 MRO）。

```mermaid
flowchart TD
    A[访问 obj.attr] --> B{实例的 __dict__<br/>里有 attr 吗?}
    B -- 有 --> C[返回实例属性]
    B -- 没有 --> D{类的 __dict__<br/>里有 attr 吗?}
    D -- 有 --> E[返回类属性]
    D -- 没有 --> F[沿 MRO 找父类]
    F --> G{找到?}
    G -- 是 --> E
    G -- 否 --> H[抛出 AttributeError]
```

````python
class Dog:
    species = "Canis familiaris"   # 类属性：所有狗共享
    count = 0                      # 类属性：统计实例数量

    def __init__(self, name):
        self.name = name           # 实例属性：每只狗独立
        Dog.count += 1             # 通过类名修改类属性


d1 = Dog("旺财")
d2 = Dog("大黄")

print(d1.species, d2.species)      # 实例可以读取类属性
print(f"共创建了 {Dog.count} 只狗")

# 关键区别：通过实例"赋值"不会修改类属性，而是创建同名实例属性！
d1.species = "狼"
print(d1.species)      # 狼  → 读到了 d1 自己的实例属性
print(d2.species)      # Canis familiaris → d2 仍然读类属性
print(Dog.species)     # Canis familiaris → 类属性没被改
print(d1.__dict__)     # 能看到 d1 多了个 species
````

## 1.7 ⚠️ 经典陷阱：可变的类属性

把**可变对象**（如列表）作为类属性，是所有新手都会踩的坑——它被**所有实例共享**：

````python
class BadDog:
    tricks = []                    # ❌ 可变类属性，所有实例共享同一个列表！

    def __init__(self, name):
        self.name = name

    def learn(self, trick):
        self.tricks.append(trick)  # 读的是类属性，append 修改了共享列表


a = BadDog("旺财")
b = BadDog("大黄")
a.learn("坐下")
print(b.tricks)   # ['坐下'] ← 大黄"莫名"学会了旺财的技能！


class GoodDog:
    def __init__(self, name):
        self.name = name
        self.tricks = []           # ✅ 实例属性，每只狗各有一份

    def learn(self, trick):
        self.tricks.append(trick)


c = GoodDog("旺财")
d = GoodDog("大黄")
c.learn("坐下")
print(c.tricks, d.tricks)   # ['坐下'] [] —— 互不影响
````

## 1.8 `__init__` 的细节

1. `__init__` **不是**创建对象的方法——真正创建对象的是 `__new__`（见第 7 章），`__init__` 只负责**初始化**已创建好的对象。
2. `__init__` 必须**没有返回值**（隐式返回 `None`），写 `return 值` 会报 `TypeError`。
3. 支持**默认参数**，让实例化更灵活：

````python
class Point:
    def __init__(self, x=0, y=0):    # 默认参数
        self.x = x
        self.y = y

    def __repr__(self):               # 第 5 章会详细讲，这里先眼熟
        return f"Point(x={self.x}, y={self.y})"


print(Point())            # 全部默认
print(Point(3))           # x=3, y 默认
print(Point(3, 4))        # 位置传参
print(Point(y=5))         # 关键字传参
````

## 1.9 对象的生命周期简述

```mermaid
flowchart LR
    A[调用 Dog 旺财] --> B[__new__<br/>分配内存 创建对象]
    B --> C[__init__<br/>初始化属性]
    C --> D[对象被使用<br/>调用方法 读写属性]
    D --> E[引用计数归零 / del]
    E --> F[__del__<br/>垃圾回收前调用]
```

- Python 通过**引用计数 + 分代回收**管理内存，通常不需要手动释放。
- `__del__` 不适合做重要清理（调用时机不确定），需要确定性清理请用**上下文管理器**（第 5 章）。
